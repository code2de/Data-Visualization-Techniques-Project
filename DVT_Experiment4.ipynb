# DVT Lab 4 — Consumer Price Index Analysis

**Dataset:** All India Consumer Price Index  
**Tools:** Python, Pandas, Matplotlib, Seaborn


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries Imported Successfully")
```

```python
# PROGRAM 1 - LOAD DATASET
df = pd.read_csv("All India Consumer Price Index.csv")

print("Dataset Loaded Successfully")
print("Dataset Shape:", df.shape)
```

```python
# PROGRAM 2 - FIRST 5 RECORDS
print("First 5 Records")
print(df.head())
```

```python
# PROGRAM 3 - LAST 5 RECORDS
print("Last 5 Records")
print(df.tail())
```

```python
# PROGRAM 4 - DATASET INFORMATION
print("Dataset Information")
df.info()
```

```python
# PROGRAM 5 - STATISTICAL SUMMARY
print("Statistical Summary")
print(df.describe(include="all"))
```

```python
# PROGRAM 6 - MISSING VALUES
print("Missing Values in Each Column")
print(df.isnull().sum())
```

```python
# PROGRAM 7 - DATA TYPES
print("Data Types")
print(df.dtypes)
```

```python
# PROGRAM 8 - UNIQUE SECTOR VALUES
print("Unique Sector Values")
print(df["Sector"].unique())
```

```python
# PROGRAM 9 - UNIQUE MONTH VALUES
print("Unique Month Values")
print(df["Month"].unique())
```

```python
# PROGRAM 10 - SECTOR-WISE RECORD COUNT
print("Sector-wise Record Count")
print(df["Sector"].value_counts())
```

```python
# PROGRAM 11 - CONVERT CPI COLUMNS TO NUMERIC
cpi_columns = [
    "Cereals and products", "Meat and fish", "Egg", "Milk and products",
    "Oils and fats", "Fruits", "Vegetables", "Pulses and products",
    "Sugar and Confectionery", "Spices", "Non-alcoholic beverages",
    "Prepared meals, snacks, sweets etc.", "Food and beverages",
    "Pan, tobacco and intoxicants", "Clothing", "Footwear",
    "Clothing and footwear", "Housing", "Fuel and light",
    "Household goods and services", "Health", "Transport and communication",
    "Recreation and amusement", "Education", "Personal care and effects",
    "Miscellaneous", "General index"
]

for col in cpi_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("CPI Columns Converted to Numeric Successfully")
print(df[cpi_columns].dtypes)
```

```python
# PROGRAM 12 - SECTOR-WISE AVERAGE GENERAL CPI
sector_average = df.groupby("Sector")["General index"].mean()

print("Average General CPI by Sector")
print(sector_average)
```

```python
# PROGRAM 13 - MAXIMUM AND MINIMUM GENERAL CPI
max_value = df["General index"].max()
min_value = df["General index"].min()

print("Maximum General CPI:", max_value)
print("Minimum General CPI:", min_value)
```

```python
# PROGRAM 14 - FILTER RURAL+URBAN RECORDS
rural_urban = df[df["Sector"] == "Rural+Urban"]

print("Rural+Urban Records")
print(rural_urban.head())
print("Number of Rural+Urban Records:", len(rural_urban))
```

```python
# PROGRAM 15 - YEAR-WISE AVERAGE GENERAL CPI
year_average = (
    df.groupby("Year")["General index"]
      .mean()
      .sort_index()
)

print("Year-wise Average General CPI")
print(year_average)
```

```python
# PROGRAM 16 - UNIQUE VALUES
print("Unique Sector Values")
print(df["Sector"].unique())

print("\nUnique Month Values")
print(df["Month"].unique())

print("\nNumber of Unique Years:", df["Year"].nunique())
```

```python
# VISUALIZATION 1 - SECTOR-WISE RECORD COUNT
sector = df["Sector"].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(sector.index, sector.values)
plt.title("Sector-wise Consumer Price Index Record Count")
plt.xlabel("Sector")
plt.ylabel("Number of Records")
plt.show()
```

```python
# VISUALIZATION 2 - AVERAGE GENERAL CPI BY SECTOR
plt.figure(figsize=(8, 5))
plt.bar(sector_average.index, sector_average.values)
plt.title("Average General CPI by Sector")
plt.xlabel("Sector")
plt.ylabel("Average General CPI")
plt.show()
```

```python
# VISUALIZATION 3 - GENERAL CPI TREND
df["Date"] = pd.to_datetime(
    df["Month"] + " " + df["Year"].astype(str),
    format="%B %Y",
    errors="coerce"
)

overall = df[df["Sector"] == "Rural+Urban"].sort_values("Date")

plt.figure(figsize=(12, 5))
plt.plot(
    overall["Date"],
    overall["General index"],
    marker="o"
)
plt.title("General CPI Trend - Rural+Urban")
plt.xlabel("Year")
plt.ylabel("General CPI")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# VISUALIZATION 4 - CORRELATION HEATMAP
heatmap_columns = [
    "Cereals and products",
    "Meat and fish",
    "Milk and products",
    "Fruits",
    "Vegetables",
    "Food and beverages",
    "Fuel and light",
    "Health",
    "Education",
    "General index"
]

plt.figure(figsize=(11, 7))
sns.heatmap(
    df[heatmap_columns].corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)
plt.title("CPI Correlation Heatmap")
plt.tight_layout()
plt.show()
```
